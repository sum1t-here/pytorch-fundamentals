```python
import torch

print(torch.__version__)
```

```python
## seeing if gpu is available or not

if torch.cuda.is_available():
    print("cuda is available")
    print(torch.cuda.get_device_name(0))
else:
    print("cuda is not available")
```

### Creating a tensor

```python
## using empty

torch.empty(2, 3)
```

```python
## type

a = torch.empty(2, 3)
type(a)
```

```python
## using 0's, 1's and rand

a = torch.zeros(2, 3)
print(a)
b = torch.ones(2, 3)
print(b)
c = torch.rand(2, 3)
print(c)
```

```python
## use of seed

torch.manual_seed(100)
a = torch.rand(2, 3)
print(a)
torch.manual_seed(100)
b = torch.rand(2, 3)
print(b)
```

```python
## using tensor

torch.tensor([[1, 2, 3],[4, 5, 6]])
```

```python
## other ways

## range 0 to 10 with 2 step
a = torch.arange(0, 10, 2)
print(a)

b = torch.linspace(0, 10, 5)
print(b)

## identity mtx
c = torch.eye(5)
print(c)

d = torch.full((2, 3), 5)
print(d)
```

```python
d.shape
```

```python
## creating a tensor like another one

torch.empty_like(d)
```

```python
torch.ones_like(d)
```

```python
torch.zeros_like(d)
```
